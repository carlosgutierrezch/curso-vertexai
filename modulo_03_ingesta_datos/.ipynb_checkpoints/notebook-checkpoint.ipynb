## 2. Crear topic y subscription de Pub/Sub

Creamos:
- **Topic principal** `hr-events`: canal donde se publican eventos de RRHH (altas, bajas, cambios salariales).
- **Topic dead letter** `hr-events-dlq`: para mensajes que fallan tras N intentos.
- **Pull subscription** `hr-events-pull-demo`: para consumir mensajes desde el notebook.

Todo idempotente: si ya existen, los reutilizamos.

```python
def get_or_create_topic(name: str):
    """Crea topic si no existe. Devuelve el path."""
    path = publisher.topic_path(PROJECT_ID, name)
    try:
        publisher.create_topic(request={"name": path})
        print(f"  Topic creado:    {name}")
    except AlreadyExists:
        print(f"  Topic ya existe: {name}")
    return path

# Crear los 4 topics que usaremos en este notebook
topic_hr_path     = get_or_create_topic(TOPIC_HR)
topic_dlq_path    = get_or_create_topic(TOPIC_DLQ)
topic_schema_path = get_or_create_topic(TOPIC_SCHEMA)
topic_bq_path     = get_or_create_topic(TOPIC_BQ)

print()
print("Inventario de topics en el proyecto:")
for t in publisher.list_topics(request={"project": f"projects/{PROJECT_ID}"}):
    print(f"  - {t.name.split('/')[-1]}")
```

```python
# --- Crear pull subscription con dead letter policy ---
sub_pull_path = subscriber.subscription_path(PROJECT_ID, SUB_PULL)

try:
    subscription = subscriber.create_subscription(
        request={
            "name": sub_pull_path,
            "topic": topic_hr_path,
            "ack_deadline_seconds": 30,
            "retry_policy": {
                "minimum_backoff": {"seconds": 5},
                "maximum_backoff": {"seconds": 60},
            },
            "dead_letter_policy": {
                "dead_letter_topic": topic_dlq_path,
                "max_delivery_attempts": 5,
            },
        }
    )
    print(f"Subscription creada: {SUB_PULL}")
except AlreadyExists:
    print(f"Subscription ya existe: {SUB_PULL}")
except PermissionDenied as e:
    print(f"  Permiso denegado para dead letter (requiere rol pubsub.publisher en el topic DLQ).")
    # Fallback: crear sin dead letter
    try:
        subscriber.create_subscription(request={"name": sub_pull_path, "topic": topic_hr_path, "ack_deadline_seconds": 30})
        print(f"Subscription creada (sin DLQ): {SUB_PULL}")
    except AlreadyExists:
        pass

# Inspeccionar
sub = subscriber.get_subscription(request={"subscription": sub_pull_path})
print(f"\nDetalles de la subscription:")
print(f"  Topic:           {sub.topic.split('/')[-1]}")
print(f"  Ack deadline:    {sub.ack_deadline_seconds}s")
print(f"  Retain acked:    {sub.retain_acked_messages}")
print(f"  Exactly-once:    {sub.enable_exactly_once_delivery}")
```

## 3. Publicar eventos sintéticos derivados de los datos reales de Personio

Leemos `dim_employee` de Silver y generamos 3 tipos de eventos realistas:
- **NUEVA_ALTA:** un empleado ficticio entra hoy
- **CAMBIO_SALARIAL:** subida del 5% para un empleado existente
- **BAJA_VOLUNTARIA:** un empleado deja la empresa

Los eventos se publican con **atributos** (event_type, country, department) que permiten filtrar en subscriptions sin parsear el body.

```python
# Cargar empleados reales para basar los eventos sintéticos en datos verosímiles
# Notas:
#   - En silver_personio.dim_employee la banda salarial está en `band` (no `position_band`).
#     La aliaseamos para mantener nombres conceptuales en el resto del notebook.
#   - Filtramos por `is_active = TRUE` (BOOL) en vez de `status = 'Active'` porque los valores
#     reales de `status` son lowercase ('active' / 'inactive' / 'onboarding').
sql_emp = f"""
SELECT
    employee_code,
    department,
    band AS position_band,
    office,
    country,
    hire_date
FROM `{PROJECT_ID}.silver_personio.dim_employee`
WHERE is_active = TRUE
LIMIT 50
"""

df_emp = bq_client.query(sql_emp).to_dataframe()
print(f"Empleados activos cargados: {len(df_emp)}")
assert len(df_emp) > 0, "df_emp está vacío — revisa el filtro o el dataset Silver"
print(df_emp.head(3).to_string(index=False))
```

```python
# --- Generar y publicar eventos sintéticos ---
import random
random.seed(42)

def publicar_evento(event_type: str, payload: dict, attrs: dict):
    """Publica un mensaje en el topic hr-events con atributos para filtrado."""
    mensaje = {
        "event_type": event_type,
        "timestamp": datetime.now(timezone.utc).isoformat(),
        "payload": payload,
    }
    data = json.dumps(mensaje, ensure_ascii=False).encode("utf-8")

    future = publisher.publish(topic_hr_path, data=data, **attrs)
    return future.result()

eventos_publicados = []

# 5 NUEVAS_ALTA (empleados ficticios)
for i in range(5):
    plantilla = df_emp.sample(1).iloc[0]
    payload = {
        "employee_code": f"EMP-DEMO-{i:03d}",
        "department": plantilla["department"],
        "position_band": plantilla["position_band"],
        "office": plantilla["office"],
        "country": plantilla["country"],
        "hire_date": "2026-04-27",
    }
    msg_id = publicar_evento("NUEVA_ALTA", payload, {
        "event_type": "NUEVA_ALTA",
        "country": str(plantilla["country"]),
        "department": str(plantilla["department"]),
    })
    eventos_publicados.append((msg_id, "NUEVA_ALTA", payload["employee_code"]))

# 3 CAMBIO_SALARIAL (empleados reales)
for _, emp in df_emp.sample(3, random_state=1).iterrows():
    payload = {
        "employee_code": emp["employee_code"],
        "department": emp["department"],
        "position_band": emp["position_band"],
        "country": emp["country"],
        "salary_change_pct": 5.0,
        "effective_date": "2026-05-01",
    }
    msg_id = publicar_evento("CAMBIO_SALARIAL", payload, {
        "event_type": "CAMBIO_SALARIAL",
        "country": str(emp["country"]),
        "department": str(emp["department"]),
    })
    eventos_publicados.append((msg_id, "CAMBIO_SALARIAL", emp["employee_code"]))

# 2 BAJA_VOLUNTARIA
for _, emp in df_emp.sample(2, random_state=2).iterrows():
    payload = {
        "employee_code": emp["employee_code"],
        "department": emp["department"],
        "country": emp["country"],
        "termination_date": "2026-04-30",
        "termination_reason": "voluntary",
    }
    msg_id = publicar_evento("BAJA_VOLUNTARIA", payload, {
        "event_type": "BAJA_VOLUNTARIA",
        "country": str(emp["country"]),
        "department": str(emp["department"]),
    })
    eventos_publicados.append((msg_id, "BAJA_VOLUNTARIA", emp["employee_code"]))

print(f"Total eventos publicados: {len(eventos_publicados)}")
print()
print(f"{'message_id':<25} {'event_type':<20} employee_code")
print("-" * 70)
for mid, etype, ecode in eventos_publicados[:10]:
    print(f"{mid:<25} {etype:<20} {ecode}")
```

## 4. Consumir mensajes con pull subscription

Usamos el **modo pull síncrono**: pedimos hasta N mensajes, los procesamos, y los confirmamos (`ack`). Patrón típico para procesamiento batch desde un notebook o un job programado.

> Si no llegan mensajes, espera unos segundos: hay latencia entre publicación y disponibilidad para pull.

```python
# Pull síncrono: pedimos hasta 20 mensajes con timeout de 5s
response = subscriber.pull(
    request={
        "subscription": sub_pull_path,
        "max_messages": 20,
    },
    timeout=10.0,
)

print(f"Mensajes recibidos: {len(response.received_messages)}\n")

ack_ids = []
for received in response.received_messages:
    msg = received.message
    body = json.loads(msg.data.decode("utf-8"))

    # Los atributos son útiles para filtrar/enrutar sin parsear el body
    attrs = dict(msg.attributes)

    print(f"  msg_id:       {msg.message_id}")
    print(f"  publish_time: {msg.publish_time}")
    print(f"  attrs:        {attrs}")
    print(f"  event_type:   {body['event_type']}")
    print(f"  employee:     {body['payload'].get('employee_code')}")
    print()

    ack_ids.append(received.ack_id)

# Confirmar (ack) para que Pub/Sub no los reentregue
if ack_ids:
    subscriber.acknowledge(request={"subscription": sub_pull_path, "ack_ids": ack_ids})
    print(f"  → {len(ack_ids)} mensajes confirmados (ack)")
```

## 5. Schema-aware topic (Avro)

Un topic con esquema **rechaza mensajes que no cumplen** la estructura definida. Útil cuando varios productores publican al mismo topic y necesitas garantizar consistencia.

Definimos un esquema Avro para `hr-events-schema` y comprobamos que un mensaje válido pasa y uno inválido falla.

```python
from google.cloud import pubsub_v1
from google.pubsub_v1.types import Schema, Encoding

schema_client = pubsub_v1.SchemaServiceClient()
schema_id = "hr-event-schema-v1"
schema_path = schema_client.schema_path(PROJECT_ID, schema_id)

# Esquema Avro para eventos HR
avro_definition = json.dumps({
    "type": "record",
    "name": "HREvent",
    "fields": [
        {"name": "event_type", "type": "string"},
        {"name": "timestamp", "type": "string"},
        {"name": "employee_code", "type": "string"},
        {"name": "country", "type": "string"},
        {"name": "department", "type": ["null", "string"], "default": None},
    ],
})

# Crear schema (idempotente)
try:
    schema = schema_client.create_schema(
        request={
            "parent": f"projects/{PROJECT_ID}",
            "schema": Schema(name=schema_path, type_=Schema.Type.AVRO, definition=avro_definition),
            "schema_id": schema_id,
        }
    )
    print(f"Schema creado: {schema.name}")
except AlreadyExists:
    print(f"Schema ya existe: {schema_id}")

# Recrear el topic con el esquema asociado (necesario en la creación)
try:
    publisher.delete_topic(request={"topic": topic_schema_path})
    print(f"Topic anterior borrado para recrear con schema")
except NotFound:
    pass

try:
    publisher.create_topic(request={
        "name": topic_schema_path,
        "schema_settings": {
            "schema": schema_path,
            "encoding": Encoding.JSON,
        },
    })
    print(f"Topic con schema creado: {TOPIC_SCHEMA}")
except AlreadyExists:
    print(f"Topic con schema ya existe: {TOPIC_SCHEMA}")
```

```python
# --- Probar: mensaje válido (debe pasar) ---
mensaje_valido = {
    "event_type": "NUEVA_ALTA",
    "timestamp": datetime.now(timezone.utc).isoformat(),
    "employee_code": "EMP-DEMO-VALID",
    "country": "ES",
    "department": "Engineering",
}
try:
    msg_id = publisher.publish(
        topic_schema_path,
        data=json.dumps(mensaje_valido).encode("utf-8")
    ).result()
    print(f"  Mensaje válido publicado: {msg_id}")
except Exception as e:
    print(f"  Error inesperado: {e}")

# --- Probar: mensaje inválido (falta employee_code, debe fallar) ---
mensaje_invalido = {
    "event_type": "NUEVA_ALTA",
    "timestamp": datetime.now(timezone.utc).isoformat(),
    "country": "ES",
    # falta employee_code (requerido por el esquema)
}
try:
    publisher.publish(
        topic_schema_path,
        data=json.dumps(mensaje_invalido).encode("utf-8")
    ).result()
    print(f"  Mensaje inválido aceptado (NO debería)")
except Exception as e:
    print(f"  Mensaje inválido rechazado (correcto):")
    print(f"     {type(e).__name__}: {str(e)[:200]}")
```

## 6. BigQuery Subscription — eventos directos a BigQuery sin código

Una **BigQuery subscription** escribe automáticamente cada mensaje en una tabla de BQ. Es la forma más simple de hacer "streaming" sin desplegar Cloud Functions ni Dataflow.

Pasos:
1. Crear la tabla destino en `bronze_personio.events_pubsub`
2. Crear una subscription tipo BigQuery sobre `hr-events-to-bq`
3. Publicar mensajes y verlos aparecer en BigQuery

```python
# 1. Crear tabla destino con el esquema esperado por BigQuery subscription
table_pubsub = f"{PROJECT_ID}.bronze_personio.events_pubsub"
schema = [
    bigquery.SchemaField("subscription_name", "STRING"),
    bigquery.SchemaField("message_id", "STRING"),
    bigquery.SchemaField("publish_time", "TIMESTAMP"),
    bigquery.SchemaField("data", "STRING"),
    bigquery.SchemaField("attributes", "STRING"),
]

table = bigquery.Table(table_pubsub, schema=schema)
table.time_partitioning = bigquery.TimePartitioning(
    type_=bigquery.TimePartitioningType.DAY, field="publish_time"
)

try:
    bq_client.create_table(table)
    print(f"Tabla creada: {table_pubsub}")
except Exception as e:
    if "Already Exists" in str(e):
        print(f"Tabla ya existe: {table_pubsub}")
    else:
        raise
```

```python
# 2. Dar permiso al service account de Pub/Sub para escribir en BigQuery
#    Una BigQuery subscription requiere que el SA de Pub/Sub tenga:
#      - roles/bigquery.dataEditor   (escribir en la tabla)
#      - roles/bigquery.metadataViewer (leer el esquema)
#    Lo concedemos a nivel de dataset (más granular que a nivel de proyecto).

# Obtener el project number (necesario para construir el email del SA de Pub/Sub)
result = subprocess.run(
    ["gcloud", "projects", "describe", PROJECT_ID, "--format=value(projectNumber)"],
    capture_output=True, text=True, check=True,
)
PROJECT_NUMBER = result.stdout.strip()
PUBSUB_SA = f"service-{PROJECT_NUMBER}@gcp-sa-pubsub.iam.gserviceaccount.com"
print(f"Project number: {PROJECT_NUMBER}")
print(f"Pub/Sub SA:     {PUBSUB_SA}")

# Grant idempotente sobre el dataset bronze_personio
from google.cloud.bigquery import AccessEntry, DatasetReference

dataset_ref = bq_client.get_dataset(f"{PROJECT_ID}.bronze_personio")
existing_emails = {e.entity_id for e in dataset_ref.access_entries if e.entity_type == "userByEmail"}

if PUBSUB_SA in existing_emails:
    print(f"  El SA de Pub/Sub ya tiene acceso al dataset bronze_personio")
else:
    nuevas_entries = list(dataset_ref.access_entries) + [
        AccessEntry(role="WRITER", entity_type="userByEmail", entity_id=PUBSUB_SA),
    ]
    dataset_ref.access_entries = nuevas_entries
    bq_client.update_dataset(dataset_ref, ["access_entries"])
    print(f"  Concedido WRITER al SA de Pub/Sub sobre bronze_personio")

# 3. Crear la BigQuery subscription
sub_bq_path = subscriber.subscription_path(PROJECT_ID, SUB_BQ)

try:
    subscription = subscriber.create_subscription(
        request={
            "name": sub_bq_path,
            "topic": topic_bq_path,
            "bigquery_config": {
                "table": table_pubsub.replace(":", "."),  # formato project.dataset.table
                "write_metadata": True,
            },
        }
    )
    print(f"\nBigQuery subscription creada: {SUB_BQ}")
except AlreadyExists:
    print(f"\nBigQuery subscription ya existe: {SUB_BQ}")
except Exception as e:
    print(f"\nNo se pudo crear la subscription: {type(e).__name__}: {str(e)[:300]}")
    print("Continuamos sin esta subscription. El resto del notebook sigue funcionando.")
```

```python
# 4. Publicar 10 mensajes en hr-events-to-bq y comprobar que aparecen en BQ
for i in range(10):
    payload = {
        "employee_code": f"EMP-BQ-{i:03d}",
        "country": random.choice(["ES", "MX", "BR", "GB"]),
        "department": random.choice(["Engineering", "Sales", "HR"]),
    }
    publisher.publish(
        topic_bq_path,
        data=json.dumps({"event_type": "DEMO_BQ", "payload": payload}).encode("utf-8"),
        event_type="DEMO_BQ",
        country=payload["country"],
    ).result()

print("10 mensajes publicados. Esperando 15s para que Pub/Sub los escriba en BQ...")
time.sleep(15)

# Consultar la tabla
sql = f"""
SELECT message_id, publish_time, attributes, SUBSTR(data, 1, 80) AS data_preview
FROM `{table_pubsub}`
WHERE publish_time > TIMESTAMP_SUB(CURRENT_TIMESTAMP(), INTERVAL 5 MINUTE)
ORDER BY publish_time DESC
LIMIT 10
"""
try:
    df = bq_client.query(sql).to_dataframe()
    print(f"\nMensajes en BQ ({len(df)} filas):")
    print(df.to_string(index=False))
except Exception as e:
    print(f"Tabla aún vacía o error: {e}")
```

## 7. Comparación práctica: batch vs streaming con datos de Personio

Simulamos los dos enfoques de ingesta y los comparamos en **coste, latencia y complejidad** usando los datos reales:

- **Batch:** carga masiva de los 209 empleados de Personio a una tabla.
- **Streaming:** publicar cada empleado como evento Pub/Sub uno a uno.

Después usamos `INFORMATION_SCHEMA.JOBS` para auditar los costes reales que generamos.

```python
# --- Enfoque BATCH: cargar 209 empleados de un golpe ---
sql_emp_full = f"""
SELECT
    employee_code,
    department,
    band AS position_band,
    office,
    country,
    hire_date
FROM `{PROJECT_ID}.silver_personio.dim_employee`
"""
df_full = bq_client.query(sql_emp_full).to_dataframe()
print(f"Empleados a procesar: {len(df_full)}")

t0 = time.time()
table_batch = f"{PROJECT_ID}.bronze_personio.demo_batch_load"

job = bq_client.load_table_from_dataframe(
    df_full,
    table_batch,
    job_config=bigquery.LoadJobConfig(write_disposition="WRITE_TRUNCATE"),
)
job.result()
t_batch = time.time() - t0

print(f"\nBATCH:")
print(f"  Filas cargadas:    {len(df_full)}")
print(f"  Tiempo total:      {t_batch:.2f}s")
print(f"  Throughput:        {len(df_full)/t_batch:.0f} filas/s")
print(f"  Coste estimado:    ~$0.00 (load jobs son gratis)")
print(f"  Tabla creada:      {table_batch}")
```

```python
# --- Enfoque STREAMING: publicar cada empleado como evento individual ---
# (Solo publicamos los primeros 30 para no saturar la subscription en clase)
N_STREAM = 30

t0 = time.time()
futures = []
for _, emp in df_full.head(N_STREAM).iterrows():
    payload = emp.to_dict()
    # to_dict puede devolver Timestamps de pandas — convertir a string
    payload = {k: (v.isoformat() if hasattr(v, "isoformat") else v) for k, v in payload.items()}

    f = publisher.publish(
        topic_hr_path,
        data=json.dumps({"event_type": "STREAM_DEMO", "payload": payload}, default=str).encode("utf-8"),
        event_type="STREAM_DEMO",
    )
    futures.append(f)

# Esperar que todos confirmen publicación
for f in futures:
    f.result()
t_stream = time.time() - t0

print(f"STREAMING:")
print(f"  Filas publicadas:  {N_STREAM}")
print(f"  Tiempo total:      {t_stream:.2f}s")
print(f"  Throughput:        {N_STREAM/t_stream:.0f} mensajes/s")

# Coste teórico Pub/Sub: $40/TiB
bytes_estimados = N_STREAM * 250
coste_pubsub = bytes_estimados / (1024**4) * 40
print(f"  Bytes publicados:  ~{bytes_estimados} bytes")
print(f"  Coste Pub/Sub:     ~${coste_pubsub:.10f} (despreciable a este volumen)")

print()
print(f"COMPARACIÓN para {N_STREAM} empleados:")
print(f"  Batch:     {t_batch:.2f}s para 209 filas (load completo)")
print(f"  Streaming: {t_stream:.2f}s para {N_STREAM} mensajes")
print(f"  Para 209 empleados streaming tardaría: ~{t_stream * 209/N_STREAM:.1f}s")
```

```python
# --- Auditar coste de los jobs de BQ generados durante este notebook ---
sql_audit = f"""
SELECT
    job_id,
    job_type,
    statement_type,
    user_email,
    creation_time,
    TIMESTAMP_DIFF(end_time, creation_time, MILLISECOND) AS duration_ms,
    total_bytes_processed,
    total_slot_ms,
    ROUND(total_bytes_processed / POW(1024, 4) * 6.25, 8) AS coste_estimado_usd
FROM `region-{REGION}`.INFORMATION_SCHEMA.JOBS_BY_USER
WHERE creation_time > TIMESTAMP_SUB(CURRENT_TIMESTAMP(), INTERVAL 30 MINUTE)
  AND state = 'DONE'
ORDER BY creation_time DESC
LIMIT 15
"""

try:
    df_jobs = bq_client.query(sql_audit).to_dataframe()
    print(f"Jobs ejecutados en los últimos 30 minutos: {len(df_jobs)}")
    if len(df_jobs):
        print()
        cols = ["job_type", "statement_type", "duration_ms", "total_bytes_processed", "coste_estimado_usd"]
        print(df_jobs[cols].head(10).to_string(index=False))
        print()
        print(f"Coste total acumulado en este notebook: ${df_jobs['coste_estimado_usd'].sum():.6f}")
except Exception as e:
    print(f"INFORMATION_SCHEMA no accesible o error: {e}")
```

## Resumen

En este notebook hemos:
1. Creado **topics y subscriptions** de Pub/Sub con dead letter policy
2. Publicado **eventos sintéticos derivados de los datos reales** de Personio (altas, cambios salariales, bajas)
3. Consumido mensajes con **pull síncrono** y confirmado con `ack`
4. Validado mensajes con **schema-aware topics** (Avro)
5. Configurado una **BigQuery Subscription** que escribe directo a BQ sin código
6. Comparado **batch vs streaming** con los 209 empleados reales de Personio
7. Auditado el coste real con `INFORMATION_SCHEMA.JOBS`

### Decisión arquitectónica para el cliente
Para 209 empleados con cambios mensuales, **batch es claramente la opción correcta**: $0 vs ~$10-50/mes en Pub/Sub para no aportar valor de negocio. Pub/Sub se justifica solo cuando se introduzca un caso de uso event-driven real (alertas, integraciones con webhooks de ATS, etc.).
